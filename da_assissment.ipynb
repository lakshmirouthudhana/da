```python
import pandas as pd

housing_df = pd.DataFrame({
    "longitude": [-118.25, -119.75, -121.50, -122.30, -117.20,
                  -120.50, -118.40, -121.80, -123.00, -116.90],

    "latitude": [34.05, 36.75, 38.58, 37.80, 32.72,
                 37.30, 34.10, 37.50, 39.00, 33.80],

    "housing_median_age": [20, 25, 30, 15, 18, 35, 22, 40, 12, 28],

    "total_rooms": [2500, 3200, 1800, 4500, 2100,
                    3800, 2900, 5000, 1600, 3400],

    "total_bedrooms": [500, 650, 400, 900, 450,
                       750, 580, 1000, 350, 700],

    "population": [1200, 1500, 900, 2200, 1100,
                   1800, 1400, 2500, 800, 1600],

    "households": [450, 600, 350, 850, 400,
                   700, 520, 950, 300, 650],

    "median_income": [3.5, 4.2, 2.8, 5.5, 3.0,
                      4.8, 3.9, 6.2, 2.5, 4.5],

    "median_house_value": [150000, 220000, 120000, 350000,
                           180000, 280000, 200000, 450000,
                           100000, 260000]
})

print(housing_df)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
print(housing_df.head())
print(housing_df.shape)
print(housing_df.info())
print(housing_df.describe())
```

Bar Chart

```python
plt.bar(housing_df.index[:10],
        housing_df["median_house_value"][:10],
        color="green")

plt.title("Housing Median House Value")
plt.xlabel("House Index")
plt.ylabel("House Value")

plt.show()
```

 Line Chart

```python
plt.figure(figsize=(10,5))

plt.plot(housing_df.index[:50],
         housing_df["median_house_value"][:50],
         color="blue",
         marker="o")

plt.title("Housing Price Trend")
plt.xlabel("House Index")
plt.ylabel("Median House Value")
plt.show()
```

Scatter Plot

```python
plt.figure(figsize=(10,5))

plt.scatter(housing_df["median_income"],
            housing_df["median_house_value"],
            color="blue",
            alpha=0.5)

plt.title("Income vs House Value")
plt.xlabel("Median Income")
plt.ylabel("Median House Value")

plt.show()
```

Pie Chart

```python
income_category = pd.cut(
    housing_df["median_income"],
    bins=3,
    labels=["Low Income", "Medium Income", "High Income"]
)

income_category.value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7,7)
)

plt.title("Housing Income Distribution")
plt.ylabel("")
plt.show()
```

Histogram

```python
plt.figure(figsize=(10,5))

plt.hist(housing_df["housing_median_age"],
         bins=20,
         color="orange",
         edgecolor="black")

plt.title("Housing Median Age Distribution")
plt.xlabel("Housing Age")
plt.ylabel("Frequency")

plt.show()
```

Box Plot

```python
plt.figure(figsize=(10,5))

plt.boxplot([
    housing_df["housing_median_age"],
    housing_df["total_rooms"],
    housing_df["total_bedrooms"]
])

plt.title("Housing Data Boxplot")
plt.xticks(
    [1, 2, 3],
    ["Housing Age", "Total Rooms", "Total Bedrooms"]
)

plt.ylabel("Values")
plt.show()
```

Area Chart

```python
plt.figure(figsize=(10,5))

plt.fill_between(
    housing_df.index[:50],
    housing_df["population"][:50],
    color="skyblue",
    alpha=0.5
)

plt.plot(
    housing_df.index[:50],
    housing_df["population"][:50]
)

plt.title("Population Area Chart")
plt.xlabel("House Index")
plt.ylabel("Population")

plt.show()
```

Donut Chart

```python
income_counts = income_category.value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    income_counts,
    labels=income_counts.index,
    autopct="%1.1f%%",
    wedgeprops={"width": 0.4}
)

plt.title("Income Category Donut Chart")
plt.show()
```

Count Chart

```python
plt.figure(figsize=(10,5))

plt.hist(
    housing_df["housing_median_age"],
    bins=10,
    color="purple",
    edgecolor="black"
)

plt.title("Housing Age Count Chart")
plt.xlabel("Housing Median Age")
plt.ylabel("Number of Houses")

plt.show()
```

Multiple Column Bar Chart

```python
housing_df[[
    "total_rooms",
    "total_bedrooms",
    "population",
    "households"
]].head(10).plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Housing Details Comparison")
plt.xlabel("House Index")
plt.ylabel("Count")
plt.legend()
plt.show()
```

Correlation Heatmap

```python
import seaborn as sns

plt.figure(figsize=(10,7))

sns.heatmap(
    housing_df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title("Housing Data Correlation Heatmap")
plt.show()
```

Geographical Scatter Chart

```python
plt.figure(figsize=(10,6))

plt.scatter(
    housing_df["longitude"],
    housing_df["latitude"],
    c=housing_df["median_house_value"],
    cmap="viridis",
    alpha=0.5
)

plt.colorbar(label="Median House Value")

plt.title("California Housing Locations")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.show()
```
